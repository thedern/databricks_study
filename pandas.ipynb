### Pandas vs Spark
Pandas is ideal for small to medium-sized datasets that fit comfortably into the memory of a single machine (the driver node in Databricks). It is generally faster for these smaller tasks due to less overhead.

Spark is designed for big data workloads (terabytes and beyond) that require processing across a distributed cluster of multiple machines. It avoids out-of-memory errors on large datasets and offers greater scalability and fault tolerance. 

### When to Use Pandas
Small datasets: Use pandas when your data can easily fit into a single machine's memory (e.g., a few gigabytes).

Exploratory Data Analysis (EDA) and Prototyping: Its intuitive API and eager execution model make it great for quick, interactive analysis and testing ideas.

Integration with other Python libraries: Pandas integrates seamlessly with the vast Python data science ecosystem, including scikit-learn and matplotlib.

Simple data manipulation: It has a lower learning curve and is often quicker to use for basic data cleaning and manipulation tasks. 

```python
"""
Using pandas in DBX
Pandas is already installed so just import and go!
"""
import pandas as pd
```

```python
"""
Dataframe
"""
data = [('John', 40, 'Journalist'), ('Mary', 32, 'Airline Pilot'), ('Dave', 13, 'Student')]

df = pd.DataFrame(data)
df.display()
```

```python
"""
Dataframe
Add column names
"""
col_names = ['name', 'age', 'occupation']
data = [('John', 40, 'Journalist'), ('Mary', 32, 'Airline Pilot'), ('Dave', 13, 'Student')]

df = pd.DataFrame(data, columns=col_names)
df.display()
```

```python
"""
Dataframe series
a 'series' is a single column from a dataframe
"""
df.age
```

```python
"""
Another notation for returning a series
Bracket notation is handy if some dummy puts a space in the column name ['my column']
"""
df['age']
```

```python
"""
get the data types from a dataframe
NOTE:  in pandas 'object' is a synonym for 'string
"""
df.dtypes
```

```python
""" 
Series operations:
series operations are simply the expressions available for that data type.  For example, the addition of int64 or concatenation of objects (strings)
"""
# add 10 to each age in the column
df.age + 10
```

```python
# concat objects (strings)
df.name + " smells like butt"
```

```python
"""
Series data has index value like python lists
"""
df.name[1]
```

```python
"""
Create a subset our dataframe and return only rows where age > 30
This is similar to spark 'select'
"""
df2 = df[df.age > 30]
df2.display()
```

```python
"""
Can pass in a python list of column names to return a subset of columns.
NOTE: The list must be inside 'brackets'
"""

df3 = df[["name", 'occupation']]
df3.display()
```

```python
"""
display as a function, not a method on the dataframe
"""
display(countries_df)
```

```python
"""
Read first five lines using the 'head' method
"""
countries_df.head(5)
```

```python
"""
Read the bottom 5 records using the 'tail' method
"""
countries_df.tail(5)

```

```python
"""
Read the first 5 lines using the 'iloc' method
"""
countries_df.iloc[:5]
```

```python
"""
Rename columns using the 'rename' method
this method takes a dictionary as an argument
"""
countries_df.rename(columns={'COUNTRY_ID': "ID"}).head(5)
```

```python
"""
Filering using a boolean T/F
"""
countries_df["REGION_ID"] == 30
```

```python
"""
Return the rows which match the filter criteria.
NOTE:  Note the dataframe + bracket notation around the filter selection
"""
countries_df[countries_df["REGION_ID"] == 30].head(5)
```

```python
"""
Filter using dot-notation
Results in the same output as above.
"""
countries_df[countries_df.REGION_ID == 30].head(5)
```

```python
"""
Filtering on multiple criteria
NOTE:  The parentheses are required and group the operands
"""
countries_df[(countries_df.REGION_ID == 30) | (countries_df.SUB_REGION_ID == 30)]
```

```python
"""
Alternate bracket syntax for the same operation as above
"""
countries_df[(countries_df["REGION_ID"] == 30) | (countries_df["SUB_REGION_ID"] == 30)]
```

```python
"""
Aggregate functions: MEAN
"""
countries_df.POPULATION.mean()
```

```python
"""
Aggregate functions: min
"""
countries_df.POPULATION.min()
```

```python
"""
Aggregate functions: max
"""
countries_df.POPULATION.max()
```

```python
"""
Dataframe summary stats with the 'describe' method
"""
countries_df.describe()
```

```python
"""
Column/Series summary stats with the 'describe' method
"""
countries_df.POPULATION.describe()
```

```python
"""
Multiple column/Series summary stats with the 'describe' method
Simply pass in the column name, no dot-notation
"""
countries_df[['POPULATION', 'AREA_KM2']].describe()
```

```python
"""
Clean up the numbers by rounding them
"""
countries_df[['POPULATION', 'AREA_KM2']].describe().round(2)
```

```python
"""
Group By
Like in spark, this create a group-ed object to which we apply aggregate functions
"""
countries_df.groupby(countries_df.REGION_ID)
```

```python
"""
Below I am grouping by region, then I am selecting the mean of the population for that specific region
"""
countries_df.groupby(countries_df.REGION_ID)['POPULATION'].mean()
```

```python
"""
Notice I need to place a list inside brackets to get multiple columns
"""
countries_df.groupby(countries_df.REGION_ID)[['POPULATION', 'AREA_KM2']].mean().round(2)
```

```python
"""
I can add a new, numeric index by using the reset_index() method
"""

countries_df.groupby(countries_df.REGION_ID)[['POPULATION', 'AREA_KM2']].mean().round(2).reset_index()
```

```python
"""
change the sort order to descending
"""

countries_df.sort_values(["POPULATION", "AREA_KM2"], ascending=False)
```

```python
"""
Sorting a column alone
"""
countries_df['POPULATION'].sort_values()
```

```python
countries_df['POPULATION'].sort_values(ascending=False)
```
