## Compute

Compute is the collection of resources available on Azure Databricks to run your notebooks, jobs, and workloads

### Serverless Compute

Serverless compute is on-demand, automatically managed compute which scales to your workload.  Automatic provisioning, configurations, and tear-down when finished

There are 3 types of serverless compute:
1. Serverless compute for notebooks.  On-demand, scalable for interactive notebooks
2. Serverless compute for jobs.   On-demand, scalable for pipeline tasks
3. Serverless compute for SQL warehouses.

### Classic Compute (all-purpose)
Explicitly provisioned compute resources which you create, configure, and modify to run your workloads.  

Can be configured for autoscaling and autotermination.

Ideal for interactive notebooks, debugging, and data experimentation

Optimized for concurrent multiple users

Not recommended for pipelines or scheduled jobs



### Jobs Compute

Ephmeral, auto-provisioned cluster type for running automated tasks and pipelines to ensure cost-efficient execution

Spin up for automated jobs, tears down when job is completed

Pay for runtime only

Optimized for automation NOT interactive use (see classic compute)

### SQL Warehouses
Compute optimized for SQL functionality.  May be Serverless or Classic in nature

## Databricks Runtimes

The runtime is the execution environment which powers compute clusters.

LTS = Long Term Support

ML = Machine Learning

The rutime consists of:
1.  Apache spark (specific version)
2.  Preinstalled libraies exeutables (scala and python)
3.  Operating system (specific version of linux)
4.  JAVA (specific version)
