## Which vendor makes the most revenue?

```python
from pyspark.sql.functions import *
```

```python
"""
table 02_silver.yellow_trips_enriched has the vendor and fare data
"""
df1 = spark.read.table('02_silver.yellow_trips_enriched')
display(df1.limit(10))
```

```python
"""
to answer this question, I can group by vendor and sum the total amound, sort decending
round the result to two decimal places
"""
df1.groupBy("vendor").\
    agg(
        round(sum("total_amount"),2).\
        alias("total_revenue")
    ).\
    orderBy("total_revenue", ascending=False).\
    display()
```

## What is the most popular pickup borough?

```python
"""
GroupBy pickup borough, count the number of records for that borough, display descending
"""
df1.groupBy('pu_borough').count().orderBy('count', ascending=False).display()
```

## What is the most common journey (borough to borough)?

```python
"""
groupBy pickup and dropoff borough

NOTE:  to use an alias for 'count', I must encapsulate the count method in an aggregation, else it throws an error.  Also when using count() in an aggregation, syntax must be count("*")

If I dont want to alias the count column, the syntax is much simpler, no agg method required
see the cell above for most popular pickup borough
"""
df1.groupBy(concat('pu_borough', lit(" -> "), 'do_borough').alias('journey')).\
    agg(
        count("*").alias('number_of_trips')
    ).\
    orderBy('number_of_trips', ascending=False).display()
```
