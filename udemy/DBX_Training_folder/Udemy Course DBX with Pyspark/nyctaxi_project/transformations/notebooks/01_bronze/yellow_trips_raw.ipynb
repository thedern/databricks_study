```python
from pyspark.sql.functions import current_timestamp
```

```python
"""
create dataframe from ALL (*) parquet files in the yellow directory
verify by displaying first 10 rows
"""
df = spark.read.parquet('/Volumes/workspace/00_landing/data_sources/nyctaxi_yellow/*')
display(df.limit(10))
```

```python
"""
transformation:
add current timestamp as new column 'processed_timestamp'
"""
df = df.withColumn("processed_timestamp", current_timestamp())
```

```python
"""
get row count
"""
df.count()
```

```python
"""
execution:
write new table with taxi cab data
"""
df.write.saveAsTable('01_bronze.yellow_trips_raw', mode='overwrite')
```

```python
"""
verify table was created successfully
"""
spark.read.table('01_bronze.yellow_trips_raw').display()
```
