```python
'''
using standard library so 'requests' does not need to be installed
'''
import urllib.request
import shutil
import os
```

```python
'''
get data from january through june, 2025
'''

months = ['01', '02', '03', '04', '05', '06']

for month in months:
    file_url = f'https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-{month}.parquet'
    dir_path = f'/Volumes/workspace/00_landing/data_sources/nyctaxi_yellow/2025-{month}'
    local_path = f'{dir_path}/yellow_tripdata_2025-{month}.parquet'
    os.makedirs(dir_path, exist_ok=True)

    binary_response = urllib.request.urlopen(file_url)
    with open(local_path, 'wb') as f:
        shutil.copyfileobj(binary_response, f)
```
