```python
'''
using standard library so 'requests' does not need to be installed
'''
import urllib.request
import shutil
import os
```

```python
'''
get data
'''

dates = ['2024-12', '2025-01', '2025-02', '2025-03', '2025-04', '2025-05']

for date in dates:
    file_url = f'https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_{date}.parquet'
    dir_path = f'/Volumes/workspace/00_landing/data_sources/nyctaxi_yellow/{date}'
    local_path = f'{dir_path}/yellow_tripdata_{date}.parquet'
    os.makedirs(dir_path, exist_ok=True)

    binary_response = urllib.request.urlopen(file_url)
    with open(local_path, 'wb') as f:
        shutil.copyfileobj(binary_response, f)
```
